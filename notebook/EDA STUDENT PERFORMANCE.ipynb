# 9. Numerical vs Categorical Features

```python
numerical_columns = [
    feature for feature in df.columns
    if df[feature].dtype != "O"
]

print("Number of Numerical Features:", len(numerical_columns))
print(numerical_columns)
```

# 10. Explore Categorical Features

```python
categorical_columns = [
    feature for feature in df.columns
    if df[feature].dtype == "O"
]

print("Number of Categorical Features:", len(categorical_columns))
print(categorical_columns)
```

# 11. Univariate Analysis

```python
# Histograms
for col in numerical_columns:
    plt.figure(figsize=(6,4))
    sns.histplot(data=df, x=col, kde=True, bins=20)

    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")

    plt.show()
```

```python
# Boxplots
for col in numerical_columns:
    plt.figure(figsize=(5,4))

    sns.boxplot(y=df[col])

    plt.title(f"Boxplot of {col}")

    plt.show()
```

```python
categorical_columns = [
    "gender",
    "race/ethnicity",
    "parental level of education",
    "lunch",
    "test preparation course"
]

for col in categorical_columns:

    plt.figure(figsize=(8,5))

    sns.countplot(data=df, x=col)

    plt.xticks(rotation=45)

    plt.title(f"Count Plot of {col}")

    plt.tight_layout()

    plt.show()
```

```python
for col in categorical_columns:

    plt.figure(figsize=(6,6))

    df[col].value_counts().plot(
        kind="pie",
        autopct="%1.1f%%"
    )

    plt.title(col)

    plt.ylabel("")

    plt.show()
```

Bivariate Analysis

```python
plt.figure(figsize=(6,4))

sns.scatterplot(
    data=df,
    x="reading score",
    y="math score"
)

plt.title("Reading Score vs Math Score")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="writing_score",
    y="math_score"
)

plt.title("Writing Score vs Math Score")

plt.show()
```

Boxplots Against Target

```python
for col in categorical_columns:

    plt.figure(figsize=(7,4))

    sns.boxplot(
        data=df,
        x=col,
        y="math score"
    )

    plt.xticks(rotation=45)

    plt.title(f"{col} vs Math Score")

    plt.show()
```

Violin Plots

```python
for col in categorical_columns:

    plt.figure(figsize=(7,4))

    sns.violinplot(
        data=df,
        x=col,
        y="math score"
    )

    plt.xticks(rotation=45)

    plt.show()
```

Mean Scores by Category

```python
for col in categorical_columns:

    print(f"\nAverage Math Score by {col}")

    print(df.groupby(col)["math score"].mean().sort_values(ascending=False))
```
